```python
import pandas as pd
import numpy as np
```

```python
expression = pd.read_csv('../data/raw/data_mrna_illumina_microarray.txt', sep='\t')

display(expression.head())
print(len(expression))
```

```python
patient_data = pd.read_parquet('../data/processed/clinical.parquet')

patient_patients = patient_data['patient_id']


expression_patients = pd.Series(expression.drop(['Hugo_Symbol', 'Entrez_Gene_Id'], axis=1).columns)


dif = expression_patients[~expression_patients.isin(patient_patients)]

expression = expression.drop(dif, axis=1)

print(f'Patients in the expression file: {len(expression_patients)}')
print(f'Patients in both tables: {len(expression_patients) - len(dif)}')
missing_mask = ~patient_patients.isin(expression_patients)
missing_ids = patient_patients[missing_mask].tolist()
print(f'Clinical patients without expression: {missing_mask.sum()} ({", ".join(missing_ids)})')
```

```python
print(f"Missing gene names: {expression['Hugo_Symbol'].isna().sum()}")
print(f"Duplicate gene names: {expression['Hugo_Symbol'].duplicated().sum()}")
```

```python
expression = expression.drop(['Entrez_Gene_Id'], axis=1)
expression = expression.groupby('Hugo_Symbol').mean()
expr_by_patient = expression.T
```

```python
assert expr_by_patient.index.is_unique
assert expr_by_patient.columns.is_unique
display(expr_by_patient.iloc[:5, :5])
print(expr_by_patient.shape)
expr_by_patient.isna().sum().sum()
```

```python
genes_with_na = expr_by_patient.columns[expr_by_patient.isna().any()]
print(f'Genes with missing values: {len(genes_with_na)}')

expr_by_patient = expr_by_patient.drop(columns=genes_with_na)
assert expr_by_patient.isna().sum().sum() == 0
```

```python
expr_by_patient.to_parquet('../data/processed/expression.parquet')
```
